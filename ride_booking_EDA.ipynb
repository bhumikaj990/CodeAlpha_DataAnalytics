```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
```

```python
booking_data=pd.read_csv("ride_booking_data.csv")
booking_data
```

```python
print(booking_data.shape)
print(booking_data.info())
print(booking_data.describe())
```

```python
booking_data['Booking_Date'] = pd.to_datetime(booking_data['Booking_Date'], 
format="%d-%m-%Y")

booking_data['Booking_Time'] = pd.to_datetime(booking_data['Booking_Time'],
 errors="coerce").dt.hour

booking_data['Fare_Amount'] = pd.to_numeric(
    booking_data['Fare_Amount'].astype(str).str.replace(r"[^\d.]", "", regex=True),
    errors="coerce"
)

booking_data.dtypes
```

```python
print(booking_data.isnull().sum())
print(booking_data.duplicated().sum())
```

```python
print(booking_data.duplicated().sum())
booking_data=booking_data.drop_duplicates()
print(booking_data.duplicated().sum())
```

```python
print(booking_data['Vehicle_Type'].unique())
print(booking_data['Payment_Method'].unique())
print(booking_data['Pickup_Location'].unique())
print(booking_data['Ride_Status'].unique())
```

```python
booking_data = booking_data[booking_data['Ride_Status'] != "Unknown"]
print(booking_data['Ride_Status'].value_counts())

print(booking_data[booking_data["Vehicle_Type"] == "Kharadi"])
booking_data = booking_data[booking_data["Vehicle_Type"] != "Kharadi"]
```

### Question 1: How many rides were completed vs. unsuccessful?

```python
counts=booking_data['Ride_Status'].value_counts()
print(counts)
```

Insight: Out of 496 bookings, about 84.5% rides were completed, while about 15.5% failed (cancelled or no driver available).

### Question 2: Which hour has the most bookings?

```python
hour_count=booking_data['Booking_Time'].value_counts().sort_index()
print(hour_count)
```

**Insight:** Bookings are spread almost evenly across the day, with about 33 to 36 rides every hour between 7 AM and 8 PM. There is no strong peak hour. The slightly lowest demand is at 8 PM (33 rides) and the highest from 8 AM to 4 PM (36 rides each).

### Question 3: Which vehicle type is used the most, and which earns the most?

```python
print(booking_data['Vehicle_Type'].value_counts())

vehical_wise_revenue=booking_data.groupby('Vehicle_Type')['Fare_Amount'].sum()
print(vehical_wise_revenue)
```

**Insight:** Mini is used the most and SUV brings the highest revenue.

### Question 4: What are the most popular pickup locations?

```python
top_pickup=booking_data['Pickup_Location'].value_counts().head(10)
print(top_pickup)
```

**Insight:** Baner and kothrud are the most popular pickup locations.

### Question 5: which payment method is most used?

```python
payment_count=booking_data['Payment_Method'].value_counts()
print(payment_count)
```

**Insight:** UPI is the most used payment method.

## Summary of Findings

**Dataset:** 500 ride booking records with 12 columns (booking date and time, ride status, vehicle type, pickup and drop locations, distance, fare, payment method, customer rating).

**Data Cleaning**
- Removed 1 record with an unknown ride status and 1 record with misaligned values (location in the Vehicle_Type column).
- Converted Fare_Amount from text to numeric.
- Checked for missing values and duplicates: 3 duplicates removed.

**Key Insights**
1. **Ride status:** 419 rides were completed, 42  were cancelled and 35 had no driver available.
2. **Peak demand:** The highest number of bookings happens around [7am to 8 pm].
3. **Vehicle type:** Mini is the most booked vehicle type (172 rides), followed by Sedan, SUV and Auto.
4. **Fare:**  SUV has the highest average fare.
5. **payment method:** UPI is the most used payment method.


**Conclusion**
Most rides are completed successfully, but about 1 in 7 bookings fails because of cancellations or driver unavailability. Improving driver availability during peak hours could reduce lost bookings.
